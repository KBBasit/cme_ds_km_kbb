```python
import requests
import pandas as pd

from pathlib import Path
import sys
project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.gdelt import fetch_data
from src.gdelt import enrich_scraped_articles
from src.scraping import scrape_articles
```

```python
response = fetch_data("Middle East conflict", "artlist", "3h")
article_list = response["articles"]
article_urls = [article["url"] for article in article_list]
```

```python
scraped_articles = scrape_articles(article_urls)
scraped_articles = enrich_scraped_articles(scraped_articles, article_list)
```

```python
df = pd.DataFrame({
    "Title": [article["title"] for article in scraped_articles],
    "URL": [article["url"] for article in scraped_articles],
    "GDELT seen date": [article["gdelt_seendate"] for article in scraped_articles],
    "Language": [article["language"] for article in scraped_articles],
    "Source country": [article["source_country"] for article in scraped_articles],
    "Text": [article["text"] for article in scraped_articles]
})

df.to_csv("../data/raw_data/gdelt_articles.csv", index=False)
```
